```python

import pandas as pd

# Create sample dataset
data = {
    "Name": ["Priya", "Arun", "Meena", "Kavin", "Ravi"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [90, 80, 95, 75, 85],
    "Department": ["AI&DS", "CSE", "IT", "AI&DS", "CSE"]
}

# Convert data into a DataFrame
df = pd.DataFrame(data)

# Save as a CSV file
df.to_csv("student_data.csv", index=False)

# Import the CSV file
df = pd.read_csv("student_data.csv")

# Display the dataset
print("CSV dataset imported successfully!")
display(df)
```

```python

print("First 5 rows:")
display(df.head())

print("Dataset shape (rows, columns):")
print(df.shape)

print("Column names:")
print(df.columns.tolist())

print("Dataset information:")
df.info()

print("Statistical summary:")
display(df.describe())
```

```python

# Check missing values
print("Missing values in each column:")
print(df.isnull().sum())

# Check duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

# Display cleaned dataset
print("\nDataset:")
display(df)
```

```python

# 1. Find the highest marks
print("Highest marks:", df["Marks"].max())

# 2. Find the lowest marks
print("Lowest marks:", df["Marks"].min())

# 3. Calculate average marks
print("Average marks:", df["Marks"].mean())

# 4. Find the student with highest marks
top_student = df.loc[df["Marks"].idxmax()]
print("\nTop student:")
display(top_student)

# 5. Department-wise average marks
print("\nDepartment-wise average marks:")
display(df.groupby("Department")["Marks"].mean())
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(df["Name"], df["Marks"])

plt.title("Student Marks Analysis")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.ylim(0, 100)

plt.show()

```
